##OpenAI embeddings


```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
embeddings
```

```python
##let us do a single text embedding\
single_text = "This is a test sentence for embedding.Langchain is great! , RAG is the future!"
single_embedding = embeddings.embed_query(single_text)
print(len(single_embedding))
print(single_embedding)
```

```python
##multiple texts at once
multiple_texts = [
    "This is the first sentence.",
    "This is the second sentence.",
    "Langchain is a powerful library for building LLM applications.",
    "RAG stands for Retrieval-Augmented Generation."
]
```

```python
multiple_embeddings = embeddings.embed_documents(multiple_texts)
```
